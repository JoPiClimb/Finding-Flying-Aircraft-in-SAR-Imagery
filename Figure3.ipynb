```python
import numpy as np
import matplotlib.pyplot as plt
```

## Azimuth displacement $d$ as a function of projected heading angle $\phi$ for different incidence angles $\theta$

```python
i_ang = np.arange(22, 48, 6)  # incidence angles [deg]
phi = np.linspace(0, 2*np.pi, 1000)

v_sat = 7.4e3      # m/s satellite speed
v_ADSB = 900/3.6   # m/s aircraft speed
H = 693            # km satellite height

phi_deg = np.rad2deg(phi)

plt.figure(figsize=(8, 5))

for theta in i_ang:

    # Approximate slant range [km]
    R_sr = H / np.cos(np.deg2rad(theta))

    # Azimuth displacement [km]
    d = (
        v_ADSB * np.sin(phi) * R_sr * np.sin(np.deg2rad(theta)) / v_sat
    )

    plt.plot(phi_deg,d,linewidth=2,label=rf'$\theta = {theta}^\circ$')

plt.xlabel(r'Projected heading angle, $\phi$ [deg]')
plt.ylabel(r'Azimuth displacement, $d$ [km]')
plt.title('')

plt.xlim(0, 360)
plt.grid(True, alpha=0.3)
plt.legend(title=r'Incidence angle')

plt.tight_layout()
plt.savefig('doppler_displacement.png', dpi=300, bbox_inches='tight')
plt.show()
```
